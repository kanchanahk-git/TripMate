```python
import pandas as pd
hotels = pd.read_csv("data/hotels.csv")
places = pd.read_csv("data/places.csv")
transport = pd.read_csv("data/transport.csv")
```

```python
import matplotlib.pyplot as plt

hotels.groupby("city")["price_per_night"].mean().plot(kind="bar")
plt.title("Average hotel price per city")
plt.ylabel("Price (₹)")
plt.show()
```

```python
hotels.groupby("category")["price_per_night"].mean().plot(kind="bar")
plt.title("Average price by category")
plt.show()
```

```python
plt.scatter(hotels["price_per_night"], hotels["rating"])
plt.xlabel("Price per night")
plt.ylabel("Rating")
plt.show()
```

```python
places["type"].value_counts().plot(kind="bar")
plt.title("Places by type")
plt.show()
```
